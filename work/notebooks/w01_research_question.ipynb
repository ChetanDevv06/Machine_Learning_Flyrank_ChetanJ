## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
from pathlib import Path

# Load the data (robust path)
data_path = Path('../../data/raw/content_refresh_anonymized.csv')
if not data_path.exists():
	print(f'Missing file: {data_path}. Run from repo root or adjust the path.')
else:
	df = pd.read_csv(data_path)
	# Calculate key statistics for CTR/Engagement Opportunity Scoring
	avg_ctr = df['ctr'].mean()
	median_engagement = df['engagement_rate'].median()
	high_traffic_low_ctr = df[(df.get('impressions_90d', 0) > 5000) & (df['ctr'] < 0.5)].shape[0]

	# Formatting: dataset stores rates as percent-scale (e.g., 0.76 == 0.76%) per data dictionary
	print(f"Average CTR across all content: {avg_ctr:.2f}%")
	print(f"Median engagement rate: {median_engagement:.2f}%")
	print(f"Pages with high traffic (>5K impressions) but low CTR (<0.5%): {high_traffic_low_ctr}")
	print('These pages represent opportunities for optimization that could significantly improve overall performance.')
```
